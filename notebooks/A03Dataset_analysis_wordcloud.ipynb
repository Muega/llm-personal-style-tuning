```python
###Part 3 
###Daten für Wordcloud vorbereiten

import datetime

class Message:
    
    def __init__(self, chatName, sender, text, when, trustlevel):
        self.name = chatName
        self.sender = sender 
        self.receiver = "Mau" if sender == chatName else chatName
        self.text = text
        self.when = datetime.datetime.strptime(when, "%d.%m.%y, %H:%M")
        self.trustlevel = trustlevel

%store -r allMessages
        
[[msg.text, msg.name, msg.sender] for msg in allMessages]
```

```python
#Emoji Optinonen für Wordcloud

import re

def keep_emoticons(text):
    #die häufigsten Emojis
    emoticons = [
        r":\)", r":D", r":P", r":d", r":p",
        r";\)", r";D", r":\(", r":O", r":o"
    ]
    
    for emo in emoticons:
        #Leerzeichen davor und dahinter  WordCloud erkennt als Wort
        text = re.sub(emo, lambda m: " " + m.group(0) + " ", text)
    
    return text

def separate_emojis(text):
    #jedes Emoji mit Leerzeichen
    return re.sub(r'([\U0001F600-\U0001F64F'
                  r'\U0001F300-\U0001F5FF'
                  r'\U0001F680-\U0001F6FF'
                  r'\U0001F1E0-\U0001F1FF])', r' \1 ', text)
def prepare_text_for_wordcloud(text):
    # 1)Emojis trennen
    text = separate_emojis(text)
    # 2)Basic Emojis :D trennen
    text = keep_emoticons(text)
    return text
```

```python
#Wordcloud plotten

import wordcloud
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
text = ' '.join([message.text for message in allMessages if message.sender=="Mau"])
clean_text = prepare_text_for_wordcloud(text)
wordcloud = WordCloud(max_font_size=60, max_words=100).generate(clean_text)
plt.figure(figsize = (10, 10))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
